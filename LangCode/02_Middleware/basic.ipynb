# LangChain Middleware – Important Points

## 🔹 Prebuilt Middleware Classes

- **SummarizationMiddleware** – Summarizes conversation when token limit is near  
- **HumanInTheLoopMiddleware** – Pauses agent for human approval  
- **ModelCallLimitMiddleware** – Limits number of model calls (cost control)  
- **ToolCallLimitMiddleware** – Limits tool execution count  
- **ModelFallbackMiddleware** – Switches to backup model if primary fails  
- **PIIMiddleware** – Detects & handles personal sensitive data  
- **TodoListMiddleware** – Adds task planning & tracking  
- **LLMToolSelectorMiddleware** – Uses LLM to select tools before main model  
- **ToolRetryMiddleware** – Retries failed tool calls automatically  
- **LLMToolEmulator** – Simulates tool execution using LLM  
- **ContextEditingMiddleware** – Trims or clears tool usage from context  
- **ShellToolMiddleware** – Provides persistent shell access  
- **FilesystemFileSearchMiddleware** – Enables file search (Glob/Grep)  
- **AgentMiddleware** – Base class for custom middleware  

---

## 🔹 Middleware Decorators

- **@before_agent** – Run logic before agent starts  
- **@before_model** – Run before each model call  
- **@after_model** – Run after model response  
- **@after_agent** – Run after agent completes  
- **@wrap_model_call** – Intercept model calls  
- **@wrap_tool_call** – Intercept tool calls  
- **@dynamic_prompt** – Create dynamic system prompts  
- **@hook_config** – Configure conditional hook behavior  

---

## 🔹 Core Types & Utilities

- **AgentState** – Stores agent execution state  
- **ModelRequest** – Data sent to model  
- **ModelResponse** – Model response data  
- **ClearToolUsesEdit** – Clears tool history from context  
- **InterruptOnConfig** – Config for human interruption  

---

## 🔹 SummarizationMiddleware Types

- **ContextSize** – Union type  
- **ContextFraction** – Summarize at context fraction  
- **ContextTokens** – Summarize at token threshold  
- **ContextMessages** – Summarize at message threshold  

# ✅ Most Important LangChain Middleware (With Simple Code Examples)

---

# 1️⃣ SummarizationMiddleware (Very Important for Chat Apps)

### ✅ Why Use?
Prevents **token overflow** in long conversations.  
Used in: Chatbots, customer support agents.

### ✅ Example

```python
from langchain.agents import create_agent
from langchain.middleware import SummarizationMiddleware
from langchain.middleware.summarization import ContextTokens

agent = create_agent(
    model="gpt-4o-mini",
    tools=[],
    middleware=[
        SummarizationMiddleware(
            context_size=ContextTokens(2000)  # Summarize when 2000 tokens reached
        )
    ]
)

# 2️⃣ ModelFallbackMiddleware (Production Safety)

## Automatically switches to backup model if main model fails.

```python
from langchain.middleware import ModelFallbackMiddleware

agent = create_agent(
    model="gpt-4o-mini",
    tools=[],
    middleware=[
        ModelFallbackMiddleware(
            fallback_models=["gpt-4o"]
        )
    ]
)

# 🚀 LangChain Middleware – Important Decorators & Core Functions (Project Quick Notes)

LangChain middleware helps you control **model calls, tool calls, prompts, logging, retries, and validation** in production agents.

---

## 🔹 @before_model
Modify input before sending to the LLM.

```python
from langchain.middleware import before_model

@before_model
def modify_input(request, state):
    request.messages.insert(0, {
        "role": "system",
        "content": "You are a helpful AI assistant."
    })
    return request



from langchain.middleware import after_model

@after_model
def process_output(response, state):
    print("Model Output:", response.content)
    return response


from langchain.middleware import wrap_model_call

@wrap_model_call
def monitor_model(call_next, request, state):
    print("Before model call")
    response = call_next(request)
    print("After model call")
    return response


from langchain.middleware import wrap_tool_call

@wrap_tool_call
def monitor_tool(call_next, request, state):
    print("Tool called:", request.tool_name)
    response = call_next(request)
    print("Tool finished")
    return response
